```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
employees=[
  (101, "  ananya rao  ", "IT", 65000.0, 28, "2024-01-15", "Chennai", None),
    (102, "KARTHIK MENON", "HR", 45000.0, 35, "2023-06-10", "Bengaluru", 5000.0),
    (103, "meera sharma", "IT", 72000.0, 31, "2022-03-20", "Hyderabad", 7000.0),
    (104, "Vikram Singh", "Finance", 58000.0, 42, "2021-11-05", "Pune", None),
    (105, "divya nair", "HR", 48000.0, 26, "2025-02-01", "Chennai", 3000.0),
    (106, "Arjun Das", "IT", 80000.0, 39, "2020-08-12", "Bengaluru", 10000.0),
    (101, "  ananya rao  ", "IT", 65000.0, 28, "2024-01-15", "Chennai", None),
]

columns=[
    "employee_id",
    "employee_name",
    "department",
    "salary",
    "age",
    "joining_date",
    "city",
    "bonus",
]
employees_df=spark.createDataFrame(data=employees, schema=columns)
employees_df.printSchema()
employees_df.show(truncate=False)
employees_df.display()
```

### Basic DataFrame operations

### Select() -- Choose Columns

```python
employees_df.select("employee_id","employee_name","salary").show(truncate=False)
```

```python
employees_df.select(F.col("employee_name"),F.col("salary")).show()
```

### alias() - Rename a result column

```python
employees_df.select(
    F.col("employee_name").alias("name"),
    F.col("salary").alias("monthly_salary"),
).show()
```

### withColumn() --Add or replace a column

```python
employees_df.withColumn("annual_saalry",F.col("salary")*12).show()
```

### withColumnRenamed() -- Rename a column

```python
renamed_df=employees_df.withColumnRenamed("employee_name","name")
renamed_df.show()
```

### drop() -- Remove columns

```python
employees_df.drop("bonus","city").show()
```

### filter()  and where()  -- select Rows

```python
employees_df.filter(F.col("salary")>50000).show()
employees_df.where(F.col("salary")>50000).show()
```

### Multiple Conditions

```python
employees_df.filter(
    (F.col("department")=="IT") & (F.col("salary")>60000)
).show()
```

### isin() --match Multiple values

```python
employees_df.filter(F.col("department").isin("IT","HR")).show()
```

### between() -- check a range 

```python
employees_df.filter(F.col("Age").between(25,35)).show()
```

### distinct() --unique rows

```python
employees_df.select("department").distinct().show()
```

### dropDuplicates() -- Remove Duplicates

```python
employees_df.dropDuplicates(["employee_id"]).show()
```

### String Functions


### trim()

```python
employees_df.select(
    "employee_name",
    F.trim("employee_name").alias("clean_name")    
).show(truncate=False)
```

### upper (), lower and initcap()

```python
employees_df.select(
    F.upper("employee_name").alias("upper_name"),
    F.lower("employee_name").alias("lower_name"),
    F.initcap("employee_name").alias("initcap_name"),
).show(truncate=False)
```

### length()

```python
employees_df.select(
    "employee_name",
    F.length(F.trim("employee_name")).alias("name_length")   
).show(truncate=False)
```

### substring()

```python
employees_df.select(
    "employee_name",
    F.substring("employee_name", 3, 5).alias("first_five_letters"),
  ).show(truncate=False)
```

### Concat() and concat_ws()

```python
display(
    employees_df.select(
        F.concat(F.lit("EMP-"), F.col("employee_id")).alias("employee_code")
    )
)

display(
    employees_df.select(
        F.concat_ws(
            " - ",
            F.col("employee_name"),
            F.col("department"),
            F.col("city")
        ).alias("Employee_details")
    )
)
```

### regex_replace()

```python
employees_df.select(
    "city",
    F.regexp_replace("city", "Chennai", "CHN").alias("city_code")
).show(truncate=False)
```

### split()

```python
employees_df.select(
    "employee_name",
    F.split(F.trim("employee_name"), " ").alias("name_parts")
).show(truncate=False)
```

### conatins() , startswith(),endswith()

```python
employees_df.filter(F.lower("employee_name").contains("meera")).show()
employees_df.filter(F.lower("city").startswith("Ch")).show()
employees_df.filter(F.lower("city").endswith("e")).show()
employees_df.filter(F.lower("city").like("%Ch%")).show()
employees_df.filter(F.lower("city").rlike(".*Ch.*")).show()
```

# Numeric Functions

### round() , ceil(),floor(), abs(),pow(),sqrt()

```python
employees_df.select(
    "salary",
    F.round(F.col("salary")/12,2).alias("monthly_component"),
    F.ceil(F.col("salary")/1000).alias("salary_thousands_up"),
    F.floor(F.col("salary")/1000).alias("salary_thousands_down")
    ).show()
```

```python
sample_numbers=spark.createDataFrame([(-10,), (5,), (-25,)], ['number'])

sample_numbers.select(
    "number",
    F.abs(F.col("number")).alias("absolute_value"),
    F.pow("number",2).alias("square")
).show()

positive_numbers=spark.createDataFrame([(4,), (9,), (16,)],["number"])
positive_numbers.select(
    "number",
    F.sqrt(F.col("number")).alias("square_root")
)
```


### Arithmetic operators

```python
employees_df.select(
    "salary",
    (F.col("salary")+5000).alias("salary_plus_5000"),
    (F.col("salary")-2000).alias("salary_minus_2000"),
    (F.col("salary")*12).alias("annual_salary"),
    (F.col("salary")/30).alias("daily_salary")
).show()
```

##  Conditional Functions

### When() and otherwise()

```python
employees_df.withColumn(
    "saalry_category",
    F.when(F.col("salary")>=70000,"High")
    .when(F.col("salary")>=50000,"Medium")
    .otherwise("Low")
).show()
```

## Null Handling

### isNull() and isNotNull()

```python
employees_df.filter(F.col("bonus").isNull()).show()
employees_df.filter(F.col("bonus").isNotNull()).show()
```

```python
employees_df.fillna({"bonus":0.0}).show()
employees_df.dropna(subset=["bonus"]).show()
```

```python
employees_df.select(
    "employee_name",
    "bonus",
    F.coalesce(F.col("bonus"),F.lit(0.0)).alias("final_bonus")
).show()
```

```python
employees_df.select(
    "employee_id",
    F.col("employee_id").cast("string").alias("text")
).printSchema()
```

DateDiff ()  and Months_between()

```python
dated_df.select(
    "employee_name",
    "joining_date",
    F.datediff(F.current_date(), F.col("joining_date")).alias("days_in_company"),
F.round(F.months_between(F.current_date(), F.col("joining_date")),1).alias("months_in_company"),
).show()
```

### add_months,date_add() and date_sub()

```python
dated_df.select(
    "joining_date",
    F.add_months("joining_date",6).alias("six_months_later"),
    F.date_add("joining_date",30).alias("30_days_later"),
    F.date_sub("joining_date",30).alias("30_days_earlier"),
  ).show()
```

### Basic Aggregations

```python
employees_df.select(
    F.count("*").alias("employee_count"),
    F.sum("salary").alias("total_salary"),
    F.round(F.avg("salary"),2).alias("average_salary"),
    F.max("salary").alias("max_salary"),
    F.min("salary").alias("min_salary"),
    F.countDistinct("department").alias("department_count"),
    F.countDistinct("bonus").alias("bonus_count"),
).show()
```

### groupby() and agg()

```python
department_report_df=(
    employees_df
    .groupBy("department")
    .agg(
        F.count("*").alias("employee_count"),
        F.sum("salary").alias("total_salary"),
        F.round(F.avg("salary"),2).alias("average_salary"),
        F.max("salary").alias("max_salary"),
        F.min("salary").alias("min_salary"
      )
    )
)
department_report_df.show()
```

# Join and union


```python
department=[
    ("IT","Technology","Chennai"),
    ("HR","Human Resources", "Bengaluru"),
    ("Finance","Finance and Accounts","Mumbai")
]

department_df=spark.createDataFrame(department,["department","name","head_office"])
joined_df=employees_df.join(department_df, on="department", how="left")
joined_df.show(truncate=False)
```


### unionByName()

```python
new_employees=[
    (107,"Rajesh","IT",50000,35,"2019-01-01","Chennai",1000),
    (108,"Priya","HR",60000,28,"2019-02-01","Bangalore",1001),
]

new_employees_df=spark.createDataFrame(new_employees,columns)
all_employees_df=employees_df.unionByName(new_employees_df)
all_employees_df.show(truncate=False)
```
